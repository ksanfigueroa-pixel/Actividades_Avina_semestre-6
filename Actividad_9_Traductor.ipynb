Se realiza la estrucura de datos que almacenara la informacion en pares en una mini traducción de algunos saludos.

```python
dic = {"hola":"hello", "buenos dias":"good morning", "buenas tardes":"good afternoon","buenas noches":"good evening", "¿como estas?":"how are you?", "¿que tal?":"how's it going?",
       "mucho gusto":"nice to meet you", "encantado":"pleased to meet you", "¿como te va?":"how are doing?", "bienvenido": "welcome"}
```

Se traduce pero si no se encuentra la palabra se envia un mensaje de descomocimiento de informacion.

```python
def tranlate (word):
  return dic.get(word, "no encontrada")
```

Se incerta la palabra que se busca traducir y se imprime la traduccion.

```python
words = input("Enter words to translate, separated by commas (e.g., hola, buenos dias): ").split(',')

for w in words:
    print(w.strip(), "->", tranlate(w.strip()))
```
